## 1. NumPy Operations

```python
# TODO: Create a NumPy array and perform vectorized operations
lst = []
for i in range(0,10):
    lst.append(i)

arr = np.array(lst)

# Arithmetic
arr * 10
arr + 5
arr ** 2

# Boolean
arr > 5
arr[arr>5]

# Aggregate Stats
int(arr.sum())
float(round(arr.std(), 3))
float(round(arr.mean(), 2))
```

## 2. Compare Loop vs Vectorization

```python
# TODO: Compare speed using %timeit

big_array = np.arange(1_000_000)

%timeit [x * 2 for x in big_array]
%timeit big_array * 2

def g(a):
    for i in a:
        i**2
%timeit g(big_array)
```
